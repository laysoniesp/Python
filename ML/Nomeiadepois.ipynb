```python
# Dataset

porco1 = [0, 0, 0]
porco2 = [1, 0, 0]
porco3 = [1, 1, 0]

gato1 = [0, 0, 1]
gato2 = [1, 0, 1]
gato3 = [1, 1, 1]

dados = [porco1, porco2, porco3, gato1, gato2, gato3]

classes = [1, 1, 1, 0, 0, 0]  # 1 = porco, 0 = gato
```

```python
# instalar scikit-learn com pip install scikit-learn

from sklearn.svm import LinearSVC

model = LinearSVC()
model.fit(dados, classes)
```
